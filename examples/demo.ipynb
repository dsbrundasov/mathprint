# mathprint: the `mpt` macro in Jupyter

Load the extension once per kernel, then use `mpt[...]` in any cell.

```python
%load_ext mathprint_jupyter
```

```python
from mathprint import mpt  # one line: enables macros and provides `mpt`
```

```python
import math as mt
import sympy as sp

a = sp.symbols("a_a")
y = 5
z = 10

# Phase 1: exact source.  Phase 2: values substituted in place.
mpt[x := y * 7 + z]
x
```

```python
# Bare expression (no assignment).
mpt[(a + z) * 2]
```

```python
# Comparison: variables are substituted in place, order preserved.
mpt[y > (a + z) * 2]
```

```python
# Module names are not expanded.
mpt[mt.floor(y) + z / 2]
```

```python
# The explicit mcpyrate form also works after loading the extension.
from mathprint_main import macros, mpt  # noqa: F401
mpt[k := z - y]
```

Unload with `%unload_ext mathprint_jupyter`.
