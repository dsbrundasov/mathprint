# mathprint: LaTeX output in Jupyter

Load the extension, then call `mpt.configure(out_format="latex")` to render `mpt` output as LaTeX (`display(Latex(py2tex(...)))`).

```python
%load_ext mathprint_jupyter
```

```python
from mathprint import mpt
mpt.configure(out_format="latex")  # <- this notebook now renders LaTeX

import sympy as sp

a = sp.symbols("a_a")
y = 5
z = 10
```

```python
# Assignment: name prefix on all three lines.
mpt[x := y * 7 + z]
x
```

```python
# Bare expression.
mpt[(a + z) * 2]
```

```python
# Comparison: values substituted in place, order preserved.
mpt[y > (a + z) * 2]
```

```python
# Back to plain text for the rest of the notebook.
mpt.configure(out_format="text")
mpt[x := y - z]
```
